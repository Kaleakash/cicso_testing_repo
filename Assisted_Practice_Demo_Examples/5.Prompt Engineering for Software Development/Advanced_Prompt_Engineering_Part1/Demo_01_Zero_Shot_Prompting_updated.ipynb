```python
# Cell 2: Import libraries and initialize the client using your Colab Secret
from google.colab import userdata
from openai import OpenAI

# Retrieve the key using your exact secret name
openai_api_key = userdata.get('OpenAI_AI_API_Key3')

# Initialize the OpenAI client
client = OpenAI(api_key=openai_api_key)

# Define the helper function for completions
def get_completion(prompt, model="gpt-4o-mini"):
    messages = [{"role": "user", "content": prompt}]
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=0,  # 0 for focused, deterministic outputs
    )
    return response.choices[0].message.content
```

```python
# Cell 3: Zero-shot prompting example
prompt = """
Classify the sentiment of the following customer review as Positive, Neutral, or Negative.
Output only the category name.

Review: "The delivery arrived ahead of schedule, but the setup manual was confusing."
Sentiment:
"""

response = get_completion(prompt)
print("Output:", response)
```

```python
!pip install -q transformers torch accelerate

import torch
from transformers import pipeline

# Runs directly in Colab memory/GPU, no API endpoints needed
generator = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    #model="meta-llama/Llama-3.2-1B-Instruct",
    device_map="auto",
    torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32
)

messages = [
    {
        "role": "user",
        "content": (
            "Classify the sentiment of the following customer review into one word: "
            "Positive, Neutral, or Negative.\n\n"
            "Review: 'The item arrived early, but the instructions were missing.'\n"
            "Sentiment:"
        )
    }
]

output = generator(messages, max_new_tokens=15, do_sample=False)
print(output[0]["generated_text"][-1]["content"].strip())

#note that- the result might be vary depending upon the models that are using
```
