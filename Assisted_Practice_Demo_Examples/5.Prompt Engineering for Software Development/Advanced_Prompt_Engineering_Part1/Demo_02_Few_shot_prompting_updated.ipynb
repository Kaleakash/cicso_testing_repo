```python
!pip install -q transformers torch accelerate

import torch
from transformers import pipeline

# Runs directly in Colab memory/GPU, no API endpoints needed
generator = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    #model="meta-llama/Llama-3.2-1B-Instruct",
    device_map="auto",
    torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32
)

messages = [
    {
        "role": "user",
        "content": (
            "Classify the sentiment of the customer review as Positive, Neutral, or Negative.\n"
            "Follow the exact format of the examples below and output only the label.\n\n"
            "Review: 'I loved the quick delivery and responsive customer support.'\n"
            "Sentiment: Positive\n\n"
            "Review: 'The package arrived on Tuesday.'\n"
            "Sentiment: Neutral\n\n"
            "Review: 'The device stopped working after just three days.'\n"
            "Sentiment: Negative\n\n"
            "Review: 'The item arrived early, but the instructions were missing.'\n"
            "Sentiment:"
        )
    }
]

output = generator(messages, max_new_tokens=15, do_sample=False)
print(output[0]["generated_text"][-1]["content"].strip())
```

In this few-shot scenario, the model uses the provided examples to understand the context and criteria for classification, applying that understanding to classify new, unseen feedback. This is particularly useful for more subjective tasks like sentiment analysis, where context and nuance play significant roles.
